```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import clear_output

# 1. Simulation Setup (1D FDTD)
Nz = 400                  # Space grid points
Nt = 450                  # Total time steps
dz = 10e-9                # Grid spacing (10 nm)
c0 = 3e8                  # Speed of light (m/s)
dt = dz / (2 * c0)        # Time step (Courant stability condition)

# Electric (Ey) aur Magnetic (Hx) fields initialize karein
Ey = np.zeros(Nz)
Hx = np.zeros(Nz)

# 2. Material Define karein (Dielectric Slab / Silicon)
# Refractive index n = 3.48 (Silicon ke liye) beech ke hisse mein rakhte hain
n = np.ones(Nz)
n[200:260] = 3.48         # Grid 200 se 260 tak Silicon slab hai
eps_r = n**2              # Relative permittivity (epsilon_r = n^2)

# Absorbing Boundary Conditions ke liye variables
ey_low_1 = ey_low_2 = 0.0
ey_high_1 = ey_high_2 = 0.0

# 3. Main Simulation Loop
for t in range(Nt):
    # Magnetic field (Hx) update karein
    Hx[:-1] += (dt / (4e-7 * np.pi * dz)) * (Ey[1:] - Ey[:-1])

    # Electric field (Ey) update karein (material ke refractive index ke saath)
    Ey[1:] += (dt / (8.854e-12 * eps_r[1:] * dz)) * (Hx[1:] - Hx[:-1])

    # Soft Gaussian Pulse Source (Light wave inject karein grid 80 par)
    pulse = np.exp(-0.5 * ((t - 60) / 15.0) ** 2)
    Ey[80] += pulse

    # Simple Absorbing Boundaries (taaki wave kinaron se takra kar wapas na aaye)
    Ey[0] = ey_low_2
    ey_low_2 = ey_low_1
    ey_low_1 = Ey[1]

    Ey[-1] = ey_high_2
    ey_high_2 = ey_high_1
    ey_high_1 = Ey[-2]

    # Har 15 steps par live graph plot karein
    if t % 15 == 0:
        clear_output(wait=True)
        plt.figure(figsize=(10, 4))

        # Silicon slab ko shaded region mein dikhayein
        plt.axvspan(200, 260, color='gray', alpha=0.3, label='Silicon Slab (n = 3.48)')

        # Electric field plot karein
        plt.plot(Ey, color='royalblue', lw=2, label='Electric Field (Ey)')

        plt.ylim(-1.2, 1.2)
        plt.xlim(0, Nz)
        plt.xlabel("Position (Grid Index)")
        plt.ylabel("Field Amplitude")
        plt.title(f"1D Photonics FDTD Simulation (Time Step: {t}/{Nt})")
        plt.legend(loc="upper right")
        plt.grid(True, linestyle='--', alpha=0.5)
        plt.show()
```
